```python
import pandas as pd
df = pd.read_csv("/content/raw-dataset.csv")
```

```python
df.shape
```

```python
import numpy as np
# Function to add artificial rows
# Function to add artificial rows
def augment_data(df, num_rows_to_add):
    augmented_data = df.copy()
    for _ in range(num_rows_to_add):
        # Randomly select a row from the existing data
        random_index = np.random.randint(0, len(df))
        random_row = df.iloc[random_index]
        # Add some noise or variation to the selected row (you can customize this)
        noisy_row = random_row.apply(lambda x: x + np.random.normal(scale=0.1) if isinstance(x, (int, float)) else x)
        # Append the noisy row to the augmented data
        augmented_data = pd.concat([augmented_data, noisy_row.to_frame().T], ignore_index=True)
    return augmented_data

# Add 100 artificial rows
augmented_df = augment_data(df, 1000)

# Now, augmented_df contains the original data plus 100 artificial rows with some variation
```

```python
from sklearn.preprocessing import LabelEncoder

Le = LabelEncoder()
List = ['Age', 'Quality of Sleep', 'Physical Activity Level', 'Stress Level', 'Heart Rate', 'Daily Steps' ]

for i in List:
  augmented_df[i] = Le.fit_transform(augmented_df[i])


```
